```python
#@title 00. Import sealed results and freeze the external-replication failure-audit protocol

from pathlib import Path
from google.colab import drive

import hashlib
import json
import platform
import sys

import numpy as np
import pandas as pd


# ============================================================
# 1. Mount Google Drive safely
# ============================================================

if Path("/content/drive/MyDrive").is_dir():
    DRIVE_ROOT = Path("/content/drive/MyDrive")

elif Path("/content/gdrive/MyDrive").is_dir():
    DRIVE_ROOT = Path("/content/gdrive/MyDrive")

else:
    drive.mount("/content/drive")
    DRIVE_ROOT = Path("/content/drive/MyDrive")


assert DRIVE_ROOT.is_dir(), DRIVE_ROOT


# ============================================================
# 2. Resolve project roots
# ============================================================

PROJECT_ROOT = (
    DRIVE_ROOT
    / "Cross-Modal_Diagnostic_Observability"
)

RETINAL_RECORD_ROOT = (
    PROJECT_ROOT
    / "06_Data_Records"
    / "Retinal_DR"
)

DEEPDRID_BASELINE_ROOT = (
    RETINAL_RECORD_ROOT
    / "Frozen_Representation_Baseline_v0.1"
)

LOCALITY_KILL_ROOT = (
    RETINAL_RECORD_ROOT
    / "Locality_Kill_Test_v0.1"
)

FINE_STRUCTURE_ROOT = (
    RETINAL_RECORD_ROOT
    / "Fine_Structure_Dissection_v0.1"
)

EXTERNAL_ROOT = (
    RETINAL_RECORD_ROOT
    / "External_Replication_Protocol_v0.1"
)

CLEAN_REPLICATION_ROOT = (
    EXTERNAL_ROOT
    / "Clean_Locality_Replication_v0.1"
)

FAILURE_AUDIT_ROOT = (
    RETINAL_RECORD_ROOT
    / "External_Replication_Failure_Audit_v0.1"
)

FAILURE_AUDIT_ROOT.mkdir(
    parents=True,
    exist_ok=True,
)


required_roots = [
    PROJECT_ROOT,
    RETINAL_RECORD_ROOT,
    DEEPDRID_BASELINE_ROOT,
    LOCALITY_KILL_ROOT,
    FINE_STRUCTURE_ROOT,
    EXTERNAL_ROOT,
    CLEAN_REPLICATION_ROOT,
]

missing_roots = [
    path
    for path in required_roots
    if not path.is_dir()
]

if missing_roots:
    raise FileNotFoundError(
        "Missing required project directories:\n"
        + "\n".join(
            str(path)
            for path in missing_roots
        )
    )


# ============================================================
# 3. Resolve sealed decision files
# ============================================================

BASELINE_DECISION_PATH = (
    DEEPDRID_BASELINE_ROOT
    / "Frozen_Representation_Baseline_v0.1_Decision.json"
)

LOCALITY_DECISION_PATH = (
    LOCALITY_KILL_ROOT
    / "Locality_Kill_Test_v0.1_Decision.json"
)

FINE_STRUCTURE_DECISION_PATH = (
    FINE_STRUCTURE_ROOT
    / "Fine_Structure_Dissection_v0.1_Final_Decision.json"
)

EXTERNAL_DECISION_PATH = (
    CLEAN_REPLICATION_ROOT
    / "EyePACS_Clean_Locality_Replication_Decision_v0.1.json"
)


required_decision_paths = [
    BASELINE_DECISION_PATH,
    LOCALITY_DECISION_PATH,
    FINE_STRUCTURE_DECISION_PATH,
    EXTERNAL_DECISION_PATH,
]

missing_decision_paths = [
    path
    for path in required_decision_paths
    if not path.is_file()
]

if missing_decision_paths:
    raise FileNotFoundError(
        "Missing sealed decision files:\n"
        + "\n".join(
            str(path)
            for path in missing_decision_paths
        )
    )


def load_json(path):
    with open(
        path,
        "r",
        encoding="utf-8",
    ) as file:
        return json.load(file)


baseline_decision = load_json(
    BASELINE_DECISION_PATH
)

locality_decision = load_json(
    LOCALITY_DECISION_PATH
)

fine_structure_decision = load_json(
    FINE_STRUCTURE_DECISION_PATH
)

external_decision = load_json(
    EXTERNAL_DECISION_PATH
)


# ============================================================
# 4. Verify the sealed decision chain
# ============================================================

assert (
    baseline_decision["decision"]
    ==
    "PASS_BASELINE_ADVANCE_TO_LOCALITY_KILL_TEST"
), baseline_decision


assert (
    locality_decision["decision"]
    ==
    "PASS_FINE_SCALE_ACHROMATIC_STRUCTURE_DEPENDENCY"
), locality_decision


assert (
    fine_structure_decision["decision"]
    ==
    "PASS_DEEPDRID_FINE_STRUCTURE_DISSECTION_"
    "ADVANCE_TO_EXTERNAL_REPLICATION"
), fine_structure_decision


assert (
    external_decision["decision"]
    ==
    "FAIL_EYEPACS_CLEAN_LOCALITY_REPLICATION_GATE"
), external_decision


# ============================================================
# 5. Freeze the interpretation boundary
# ============================================================

interpretation_boundary = {
    "audit_status": (
        "EXPLORATORY_AFTER_EXTERNAL_VALIDATION_UNBLINDING"
    ),
    "confirmed_fact": (
        "The sealed DeepDRiD eye-local severity-direction "
        "result did not replicate on the sealed EyePACS "
        "validation cohort."
    ),
    "prohibited_claim": (
        "No analysis performed in this failure audit may "
        "be presented as an independent EyePACS confirmation."
    ),
    "prohibited_actions": [
        (
            "Do not redefine the external replication gate."
        ),
        (
            "Do not tune model hyperparameters using the "
            "EyePACS validation labels and then call the "
            "result confirmatory."
        ),
        (
            "Do not select a favourable EyePACS subgroup "
            "post hoc and describe it as external replication."
        ),
        (
            "Do not overwrite or replace any sealed decision "
            "from the preceding notebooks."
        ),
    ],
    "permitted_goal": (
        "Use exploratory analyses to identify plausible, "
        "measurable reasons for replication failure and to "
        "design a genuinely prospective retinal blind test."
    ),
}


# ============================================================
# 6. Freeze failure-audit protocol
# ============================================================

protocol_rows = [
    {
        "stage": 0,
        "analysis": (
            "Sealed decision-chain import"
        ),
        "status": "pre-registered",
        "evidence_role": (
            "provenance and interpretation control"
        ),
        "primary_question": (
            "Are all discovery, mechanism and external "
            "replication decisions imported without alteration?"
        ),
        "permitted_information": (
            "Previously sealed decision files only"
        ),
        "checkpoint": (
            "Proceed only if all four decisions match exactly."
        ),
    },
    {
        "stage": 1,
        "analysis": (
            "Cohort and label-structure audit"
        ),
        "status": "pre-registered",
        "evidence_role": (
            "exploratory failure explanation"
        ),
        "primary_question": (
            "Do DeepDRiD and EyePACS differ in grade-gap, "
            "grade-pair, direction, prevalence or sample "
            "structure?"
        ),
        "permitted_information": (
            "DeepDRiD and EyePACS metadata and labels"
        ),
        "checkpoint": (
            "Review whether the task is genuinely comparable "
            "before interpreting representation failure."
        ),
    },
    {
        "stage": 2,
        "analysis": (
            "Label-free acquisition and embedding domain shift"
        ),
        "status": "pre-registered",
        "evidence_role": (
            "candidate CDO target-support evidence"
        ),
        "primary_question": (
            "Can DeepDRiD and EyePACS be distinguished using "
            "image statistics or frozen embeddings without "
            "using DR labels?"
        ),
        "permitted_information": (
            "Images, acquisition statistics and embeddings; "
            "no DR labels in domain models"
        ),
        "checkpoint": (
            "Assess whether replication failure was accompanied "
            "by prospectively observable target mismatch."
        ),
    },
    {
        "stage": 3,
        "analysis": (
            "Direct and reverse diagnostic-axis transfer"
        ),
        "status": "pre-registered",
        "evidence_role": (
            "mechanism discrimination"
        ),
        "primary_question": (
            "Does the DeepDRiD diagnostic axis transfer to "
            "EyePACS, and does the EyePACS axis transfer back "
            "to DeepDRiD?"
        ),
        "permitted_information": (
            "Sealed embeddings, frozen probes and held-out "
            "predictions"
        ),
        "checkpoint": (
            "Separate representation-domain mismatch from "
            "diagnostic-axis instability."
        ),
    },
    {
        "stage": 4,
        "analysis": (
            "Grade-gap and grade-pair heterogeneity audit"
        ),
        "status": "pre-registered",
        "evidence_role": (
            "exploratory boundary-condition analysis"
        ),
        "primary_question": (
            "Is observability concentrated in particular "
            "severity gaps or grade combinations?"
        ),
        "permitted_information": (
            "Labels may be used, but all conclusions remain "
            "exploratory"
        ),
        "checkpoint": (
            "Do not promote a post-hoc subgroup to independent "
            "replication."
        ),
    },
    {
        "stage": 5,
        "analysis": (
            "Baseline comparison and failure synthesis"
        ),
        "status": "pre-registered",
        "evidence_role": (
            "retinal feasibility decision"
        ),
        "primary_question": (
            "Do information-support variables explain failure "
            "beyond simple quality and embedding-distance "
            "baselines?"
        ),
        "permitted_information": (
            "Outputs from stages 1–4"
        ),
        "checkpoint": (
            "Issue continue, narrow-pivot or stop decision."
        ),
    },
    {
        "stage": 6,
        "analysis": (
            "Prospective retinal blind-test specification"
        ),
        "status": "conditional",
        "evidence_role": (
            "future confirmatory test"
        ),
        "primary_question": (
            "Can a new target outcome and failure mechanism be "
            "predicted before target labels are revealed?"
        ),
        "permitted_information": (
            "Unlabelled target images and pre-frozen source "
            "evidence only"
        ),
        "checkpoint": (
            "Run only if failure audit yields a measurable and "
            "falsifiable target-support hypothesis."
        ),
    },
]


failure_audit_protocol = pd.DataFrame(
    protocol_rows
)


# ============================================================
# 7. Fingerprint sealed inputs
# ============================================================

def sha256_file(
    path,
    chunk_size=1024 * 1024,
):
    digest = hashlib.sha256()

    with open(
        path,
        "rb",
    ) as file:
        while True:
            chunk = file.read(
                chunk_size
            )

            if not chunk:
                break

            digest.update(chunk)

    return digest.hexdigest()


sealed_input_manifest = pd.DataFrame(
    [
        {
            "artifact": (
                "DeepDRiD frozen baseline decision"
            ),
            "path": str(
                BASELINE_DECISION_PATH
            ),
            "sha256": sha256_file(
                BASELINE_DECISION_PATH
            ),
            "decision": (
                baseline_decision["decision"]
            ),
        },
        {
            "artifact": (
                "DeepDRiD locality kill-test decision"
            ),
            "path": str(
                LOCALITY_DECISION_PATH
            ),
            "sha256": sha256_file(
                LOCALITY_DECISION_PATH
            ),
            "decision": (
                locality_decision["decision"]
            ),
        },
        {
            "artifact": (
                "DeepDRiD fine-structure decision"
            ),
            "path": str(
                FINE_STRUCTURE_DECISION_PATH
            ),
            "sha256": sha256_file(
                FINE_STRUCTURE_DECISION_PATH
            ),
            "decision": (
                fine_structure_decision["decision"]
            ),
        },
        {
            "artifact": (
                "EyePACS clean external-replication decision"
            ),
            "path": str(
                EXTERNAL_DECISION_PATH
            ),
            "sha256": sha256_file(
                EXTERNAL_DECISION_PATH
            ),
            "decision": (
                external_decision["decision"]
            ),
        },
    ]
)


# ============================================================
# 8. Save protocol artifacts
# ============================================================

PROTOCOL_PATH = (
    FAILURE_AUDIT_ROOT
    / (
        "External_Replication_Failure_Audit_"
        "Protocol_v0.1.csv"
    )
)

INTERPRETATION_BOUNDARY_PATH = (
    FAILURE_AUDIT_ROOT
    / (
        "External_Replication_Failure_Audit_"
        "Interpretation_Boundary_v0.1.json"
    )
)

SEALED_INPUT_MANIFEST_PATH = (
    FAILURE_AUDIT_ROOT
    / (
        "External_Replication_Failure_Audit_"
        "Sealed_Input_Manifest_v0.1.csv"
    )
)

BOOTSTRAP_PATH = (
    FAILURE_AUDIT_ROOT
    / (
        "External_Replication_Failure_Audit_"
        "Bootstrap_v0.1.json"
    )
)


failure_audit_protocol.to_csv(
    PROTOCOL_PATH,
    index=False,
)

sealed_input_manifest.to_csv(
    SEALED_INPUT_MANIFEST_PATH,
    index=False,
)


with open(
    INTERPRETATION_BOUNDARY_PATH,
    "w",
    encoding="utf-8",
) as file:
    json.dump(
        interpretation_boundary,
        file,
        indent=2,
    )


bootstrap_record = {
    "project": (
        "Cross-Modal Diagnostic Observability"
    ),
    "analysis": (
        "Retinal external-replication failure audit"
    ),
    "version": "v0.1",
    "status": (
        "EXPLORATORY_AFTER_EXTERNAL_VALIDATION_UNBLINDING"
    ),
    "deepdrid_baseline_decision": (
        baseline_decision["decision"]
    ),
    "deepdrid_locality_decision": (
        locality_decision["decision"]
    ),
    "deepdrid_fine_structure_decision": (
        fine_structure_decision["decision"]
    ),
    "eyepacs_external_replication_decision": (
        external_decision["decision"]
    ),
    "deepdrid_primary_accuracy": (
        0.6888888888888889
    ),
    "deepdrid_primary_directional_auc": (
        0.7530364372469636
    ),
    "eyepacs_validation_accuracy": (
        external_decision[
            "validation_pairwise_accuracy"
        ]
    ),
    "eyepacs_validation_directional_auc": (
        external_decision[
            "validation_directional_auc"
        ]
    ),
    "eyepacs_validation_one_sided_p": (
        external_decision[
            "validation_exact_one_sided_p"
        ]
    ),
    "python": sys.version,
    "platform": platform.platform(),
    "numpy": np.__version__,
    "pandas": pd.__version__,
}


with open(
    BOOTSTRAP_PATH,
    "w",
    encoding="utf-8",
) as file:
    json.dump(
        bootstrap_record,
        file,
        indent=2,
    )


# ============================================================
# 9. Final report
# ============================================================

print(
    "================ FAILURE-AUDIT "
    "BOOTSTRAP ================"
)

print(
    "DeepDRiD baseline:"
)

print(
    baseline_decision["decision"]
)

print(
    "\nDeepDRiD locality kill test:"
)

print(
    locality_decision["decision"]
)

print(
    "\nDeepDRiD fine-structure dissection:"
)

print(
    fine_structure_decision["decision"]
)

print(
    "\nEyePACS clean replication:"
)

print(
    external_decision["decision"]
)


print(
    "\nFailure-audit output root:"
)

print(
    FAILURE_AUDIT_ROOT
)


print(
    "\nInterpretation status:"
)

print(
    interpretation_boundary[
        "audit_status"
    ]
)


print(
    "\nFrozen protocol:"
)

display(
    failure_audit_protocol
)


print(
    "\nSealed inputs:"
)

display(
    sealed_input_manifest[
        [
            "artifact",
            "decision",
            "sha256",
        ]
    ]
)


print(
    "\nSaved:"
)

print(
    PROTOCOL_PATH
)

print(
    INTERPRETATION_BOUNDARY_PATH
)

print(
    SEALED_INPUT_MANIFEST_PATH
)

print(
    BOOTSTRAP_PATH
)


print(
    "\nExternal-replication failure-audit "
    "protocol frozen successfully."
)
```
